Para entender como computadores representam textos, podemos olhar para a tabela [ASCII](https://pt.wikipedia.org/wiki/ASCII):
- A função `ord(c)` retorna o código numérico de um caractere. Exemplo: `ord('a')` é 97 e `ord('A')` é 65.
- A função `chr(n)` faz o caminho inverso, transformando o número de volta em caractere.
- A diferença entre qualquer letra minúscula ('a' a 'z') e sua equivalente maiúscula ('A' a 'Z') é exatamente 32 posições numéricas: `ord('a') - 32` resulta em 65, que é `ord('A')`.

Dica: veja a demonstração abaixo.

```python
# Demonstração do funcionamento de ord e chr
codigo_minisculo = ord("b")
codigo_maiusculo = codigo_minisculo - 32
letra_maiuscula = chr(codigo_maiusculo)

print("Caractere original:", "b")
print("Código numérico:", codigo_minisculo)
print("Novo código numérico (-32):", codigo_maiusculo)
print("Caractere resultante:", letra_maiuscula)
```
